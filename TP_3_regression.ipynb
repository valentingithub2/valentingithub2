# TP n°3 Prédiction du taux de nitrogène : cas  de la regression dans le Machine Learning 

Introduction

Nitrogène ou Azote en francais est un élément nutritif essenciel pour les plantes.
Il est un composant clés des protéines, enzymes et chlorophylles 
L'azote est un élement sous differentes formes dans le sol (nitrates, ammonium) ou dans la matière organique.

Il s'agit de mettre en place un algorithme de prédiction de la quantité de nitrates en fonction d'autres paramètres du sol


```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import plotly.express as px
import seaborn as sns 
```

```python
#importation des données
df = pd.read_csv('D:/Downloads/ACADEMIE/fichier_fertilisation.csv', sep=';')
df.head()
```

```python
#selectionner la variable cible Y 
Y = df['Nitrogen']
X = df.drop(columns=['Nitrogen', 'Soil Type', 'Crop Type', 'Fertilizer Name'])
```

```python
#faire la procédure Train-Test-Split
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state=5)
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

lmodellinaire = LinearRegression()
lmodellinaire.fit(X_train, Y_train)

#evaluation du trainning set 
from sklearn.metrics import r2_score
y_train_predict = lmodellinaire.predict(X_train)
rmse = (np.sqrt(mean_squared_error(Y_train, y_train_predict)))
r2 = r2_score(Y_train, y_train_predict)

print('La performance du modèle sur la base dapprentissage est')
print('--------------------------------------')
print('Lerreur quadratique moyenne est {}'.format(rmse))
print('le score R2 est {}'.format(r2))



y_test_predict = lmodellinaire.predict(X_test)
rmse = (np.sqrt(mean_squared_error(Y_test, y_test_predict)))
r2 = r2_score(Y_test, y_test_predict)

print('La performance du modèle sur la base de test')
print('-------------------------------------')
print('lerreur quadratique moyenne est {}'.format(rmse))
print('le score R2 est {}'.format(r2))

```

```python
#affichage eds coeffcients de chaque variable 
coef = pd.concat([pd.Series(X_train.columns), pd.Series(lmodellinaire.coef_)], axis=1)
coef.columns=["variable", "coefficient"]
print(coef)
print('Intercept: ', lmodellinaire.intercept_)
```

```python
#algorithme d'arbre de decision 
from sklearn.tree import DecisionTreeRegressor, plot_tree
tree = DecisionTreeRegressor() 
tree.fit(X_train, Y_train)
print(tree.get_depth())
y_pred_train = tree.predict(X_train)
y_pred_test = tree.predict(X_test)

print("RMSE sur train :", mean_squared_error(Y_train, y_pred_train, squared=False))
print("RMSE sur test :", mean_squared_error(Y_test, y_pred_test, squared=False))
print('----------------------------------------')
print("R² sur train :", r2_score(Y_train, y_pred_train))
print("R² sur test :", r2_score(Y_test, y_pred_test))
```

```python
from sklearn.ensemble import RandomForestRegressor
tree = RandomForestRegressor(random_state=42)
tree.fit(X_train, Y_train)
y_pred_train = tree.predict(X_train)
y_pred_test = tree.predict(X_test)

print("RMSE sur train :", mean_squared_error(Y_train, y_pred_train, squared=False))
print("RMSE sur test :", mean_squared_error(Y_test, y_pred_test, squared=False))
print('----------------------------------------------')
print("R² sur train :", r2_score(Y_train, y_pred_train))
print("R² sur test :", r2_score(Y_test, y_pred_test))
```
